### Task 2: A Method That Returns a Value

Create two classes, `Circle` (with `radius`) and `Square` (with `side`), each with an `area()` method (`radius ** 2 * 3.1416` and `side ** 2` respectively). Put one of each into a list and loop through it, printing each shape's area — this is polymorphism in action.

```python
# --- Task 6: A Method That Uses Other Attributes ---
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

rect1 = Rectangle(5, 3)
print("Rectangle Area:", rect1.area())
print("Rectangle Perimeter:", rect1.perimeter())
```

```python
# --- Task 7: Encapsulation with a Guard Condition ---
class Wallet:
    def __init__(self, balance):
        self._balance = balance

    def withdraw(self, amount):
        if amount <= self._balance:
            self._balance -= amount
            print(f"Withdrew {amount}. New balance: {self._balance}")
        else:
            print("Insufficient funds.")

    def get_balance(self):
        return self._balance

print("\nWallet Test:")
wallet1 = Wallet(1000)
wallet1.withdraw(300)     # succeeds
wallet1.withdraw(5000)    # fails
print("Final Balance:", wallet1.get_balance())
```

```python
# --- Task 8: Basic Inheritance ---
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def describe(self):
        return f"{self.brand} can travel at {self.speed} km/h."

class ElectricVehicle(Vehicle):
    def __init__(self, brand, speed, battery_range):
        super().__init__(brand, speed)
        self.battery_range = battery_range

    def describe(self):
        base = super().describe()
        return f"{base} It has a battery range of {self.battery_range} km."

ev = ElectricVehicle("Tesla", 250, 500)
print("\nEV Description:", ev.describe())
```

```python
# --- Task 9: A List of Mixed Objects with a Shared Method ---
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return self.radius ** 2 * 3.1416

class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

shapes = [Circle(5), Square(4)]
print("\nPolymorphism Area Test:")
for shape in shapes:
    print(f"{type(shape).__name__} area: {shape.area():.2f}")
```

```python
# --- Task 10: Combine Class + List Comprehension ---
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

products = [
    Product("Keyboard", 45),
    Product("Monitor", 150),
    Product("Mouse", 25),
    Product("Laptop", 1200),
]

expensive_names = [p.name for p in products if p.price > 100]
print("\nProducts priced above 100:", expensive_names)
```

### Task 13: Records as Objects
You previously worked with student records as dictionaries: `{"name": "Ali", "marks": [78, 85, 90]}`. Rewrite this idea using a class `StudentRecord` with `name` and `marks` attributes, and methods `average()` and `grade()` (same grading rule as earlier notebooks: `>=90` → `"A"`, `>=75` → `"B"`, `>=60` → `"C"`, else `"F"`). Create a list of at least 4 `StudentRecord` objects, then use a list comprehension to build a list of names of students with grade `"A"` or `"B"`.

```python
# --- Task 11: A Simple Dataset Class ---
class Dataset:
    def __init__(self, values):
        self.values = values

    def mean(self):
        return sum(self.values) / len(self.values)

    def minimum(self):
        return min(self.values)

    def maximum(self):
        return max(self.values)

    def normalize(self):
        low = self.minimum()
        high = self.maximum()
        return [(v - low) / (high - low) for v in self.values]

data = Dataset([10, 20, 30, 40, 50])
print("Dataset Mean:", data.mean())
print("Dataset Min:", data.minimum())
print("Dataset Max:", data.maximum())
print("Dataset Normalized:", data.normalize())
```

```python
# --- Task 12: A Base Model Class and Two Subclasses ---
class BaseModel:
    def __init__(self):
        self.is_trained = False

    def predict(self):
        if not self.is_trained:
            raise ValueError("Model has not been trained yet. Call fit() first.")

class ConstantModel(BaseModel):
    def __init__(self):
        super().__init__()
        self.constant = None

    def fit(self, training_values):
        self.constant = max(training_values)
        self.is_trained = True

    def predict(self, n):
        super().predict()
        return [self.constant] * n

class AverageModel(BaseModel):
    def __init__(self):
        super().__init__()
        self.average = None

    def fit(self, training_values):
        self.average = sum(training_values) / len(training_values)
        self.is_trained = True

    def predict(self, n):
        super().predict()
        return [self.average] * n

training_data = [5, 10, 15, 20, 25]

constant_model = ConstantModel()
constant_model.fit(training_data)
print("\nConstantModel predictions:", constant_model.predict(3))

average_model = AverageModel()
average_model.fit(training_data)
print("AverageModel predictions:", average_model.predict(3))

untrained_model = AverageModel()
try:
    untrained_model.predict(2)
except ValueError as error:
    print("Caught expected error for untrained model:", error)
```

```python
# --- Task 13: Records as Objects ---
class StudentRecord:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks) / len(self.marks)

    def grade(self):
        avg = self.average()
        if avg >= 90:
            return "A"
        elif avg >= 75:
            return "B"
        elif avg >= 60:
            return "C"
        else:
            return "F"

students = [
    StudentRecord("Ali", [78, 85, 90]),
    StudentRecord("Sara", [88, 92, 95]),
    StudentRecord("Ahmed", [55, 62, 58]),
    StudentRecord("Ayesha", [90, 87, 93]),
]

top_students = [s.name for s in students if s.grade() in ("A", "B")]
print("\nStudents with grade A or B:", top_students)
```
